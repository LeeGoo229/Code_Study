## 문제 1. 상품 카테고리별 주문 건수 막대그래프 [필수 기초]

`online`에서 `product_category`별 주문 건수를 집계한 뒤, 카테고리를 X축, 주문 건수를 Y축으로 하는 수직 막대그래프(`plt.bar`)를 그리세요. 제목은 `Orders by Product Category`로 지정하세요.

**점검 기준:** 온라인 주문 12건 중 `Tech` 5건, `Food` 4건, `Book` 3건이며 `Tech` 막대가 가장 높습니다.

````python
# 문제 1 풀이
cat_counts = online['product_category'].value_counts()

plt.figure(figsize=(7, 4))
plt.bar(cat_counts.index, cat_counts.values, color='#3498db', edgecolor='white', width=0.55)
plt.title('Orders by Product Category')
plt.xlabel('Category')
plt.ylabel('Order Count')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()

print("온라인 카테고리별 주문 건수:")
print(cat_counts)
````

**결과 해석 / 근거 / 새로 알게 된 점:**

1. **카테고리별 주문 건수:** 온라인 전체 12건 중 `Tech`가 5건(41.7%)으로 가장 많고, `Food`가 4건(33.3%), `Book`이 3건(25.0%)으로 집계되어 Tech 막대가 가장 높게 나타납니다.
2. 막대그래프는 이산적인 범주형 데이터의 빈도 크기를 시각적 막대 높이로 명확히 서열화하여 비교하는 데 가장 효과적입니다.


## 문제 2. 지역별 매출 합계 막대그래프 [필수 기초]

`online`과 `offline`을 행 방향으로 결합한 `all_orders`(총 20행)를 만들고, `region`별 `revenue` 합계를 계산해 막대그래프로 비교하세요.

**점검 기준:** 지역별 매출 합계는 `South` 2550, `North` 750, `East` 750, `West` 370으로 `South` 지역의 매출이 가장 큽니다(전체 합계 4420).

````python
# 문제 2 풀이
all_orders = pd.concat([online, offline], ignore_index=True)
region_rev = all_orders.groupby('region')['revenue'].sum().loc[['South', 'North', 'East', 'West']]

plt.figure(figsize=(7, 4))
plt.bar(region_rev.index, region_rev.values, color='#2ecc71', edgecolor='white', width=0.55)
plt.title('Total Revenue by Region (Online + Offline)')
plt.xlabel('Region')
plt.ylabel('Total Revenue')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()

print("지역별 매출 합계:")
print(region_rev)
print(f"전체 합계: {region_rev.sum():,}원")
````

**결과 해석 / 근거 / 새로 알게 된 점:**

1. **온·오프라인 통합 20행 집계 결과:**
   - `South`: **2,550원** (전체 매출의 57.7%를 차지하는 압도적 1위)
   - `North`: **750원**
   - `East`: **750원**
   - `West`: **370원**
   - 전체 총매출 합계: **4,420원**으로 점검 기준과 정확히 일치합니다.
2. 온·오프라인 채널을 결합하여 전체 거시적 관점에서 South 권역이 회사의 주력 핵심 상권임을 명확히 규명했습니다.


## 문제 3. 막대그래프(`plt.bar`)와 히스토그램(`plt.hist`) 나란히 비교하기 [필수 기초]

`1행 2열` 서브플롯에 왼쪽은 문제 1의 카테고리별 주문 건수(`plt.bar`), 오른쪽은 `online['revenue']`의 분포(`plt.hist(bins=5, edgecolor='white')`)를 그리고 두 차트의 X축 의미와 막대 간격 차이를 설명하세요.

**점검 기준:** 왼쪽 `bar` 차트는 독립된 범주(`Tech`, `Food`, `Book`) 사이에 간격이 있고, 오른쪽 `hist`는 연속된 매출 금액 구간이라 막대가 빈틈없이 붙어 있습니다.

````python
# 문제 3 풀이
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

# 왼쪽: 카테고리별 주문 건수 (plt.bar)
cat_counts = online['product_category'].value_counts()
axes[0].bar(cat_counts.index, cat_counts.values, color='#3498db', width=0.5)
axes[0].set_title('Orders by Category (plt.bar)')
axes[0].set_xlabel('Product Category (범주형)')
axes[0].set_ylabel('Order Count')
axes[0].grid(axis='y', linestyle='--', alpha=0.4)

# 오른쪽: 매출 분포 (plt.hist)
axes[1].hist(online['revenue'], bins=5, edgecolor='white', color='#e74c3c')
axes[1].set_title('Revenue Distribution (plt.hist, bins=5)')
axes[1].set_xlabel('Revenue (연속형 수치)')
axes[1].set_ylabel('Frequency')
axes[1].grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
````

**결과 해석 / 근거 / 새로 알게 된 점:**

### 막대그래프(Bar) vs 히스토그램(Hist)의 근본적 차이
1. **X축의 데이터 특성:**
   - 왼쪽 `bar` 차트: 독립적인 이산형 범주(`Tech`, `Food`, `Book`)를 다루며 순서가 임의로 바뀔 수 있습니다.
   - 오른쪽 `hist` 차트: 연속적인 수치형 금액 구간(70 ~ 500원)을 다루며 순서를 바꿀 수 없습니다.
2. **막대 사이의 간격:**
   - `bar` 차트는 서로 다른 범주임을 명시하기 위해 막대 사이에 빈 공간(Gap)이 존재합니다.
   - `hist` 차트는 수치가 끊어짐 없이 연속적으로 이어진다는 것을 표현하기 위해 막대들이 빈틈없이 서로 붙어 있습니다.


## 문제 4. 상품 카테고리별 매출 내림차순 정렬 막대그래프 [필수 기초]

`online`에서 `product_category`별 `revenue` 합계를 구하고 매출이 높은 순서(`ascending=False`)로 정렬한 뒤 수직 막대그래프를 그리세요.

**점검 기준:** 내림차순 정렬 결과는 `Tech(2050)` → `Book(400)` → `Food(350)` 순이며 가장 왼쪽에 `Tech`, 그다음 `Book`, `Food` 막대가 표시됩니다.

````python
# 문제 4 풀이
cat_rev_sorted = online.groupby('product_category')['revenue'].sum().sort_values(ascending=False)

plt.figure(figsize=(7, 4))
plt.bar(cat_rev_sorted.index, cat_rev_sorted.values, color='#f39c12', edgecolor='white', width=0.55)
plt.title('Online Revenue by Category (Descending)')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()

print("카테고리별 매출 내림차순:")
print(cat_rev_sorted)
````

**결과 해석 / 근거 / 새로 알게 된 점:**

1. **내림차순 정렬 결과:** **`Tech(2,050원)` -> `Book(400원)` -> `Food(350원)`** 순서로 배치되었습니다.
2. 정렬되지 않은 임의 순서의 차트에 비해, 매출이 높은 순서로 좌에서 우로 계단식 하강 정렬함으로써 어떤 상품군이 1위, 2위, 3위인지 한눈에 인지할 수 있어 시각적 전달력이 비약적으로 향상됩니다.


## 문제 5. 긴 범주명을 위한 수평 막대그래프(`plt.barh`)와 정렬 [필수 기초]

`inquiries` 데이터를 문의 건수(`count`) 기준으로 정렬하여 가장 건수가 많은 항목이 **맨 위**에 오도록 `plt.barh()` 수평 막대그래프를 그리세요.

**점검 기준:** `ascending=True`로 정렬해야 `plt.barh`에서 맨 아래에 `회원 등급 및 포인트 적립 문의(20)`가, 맨 위에 `상품 배송 일정 및 지연 문의(45)`가 배치되어 읽기 편합니다.

````python
# 문제 5 풀이
inq_sorted = inquiries.sort_values(by='count', ascending=True)

plt.figure(figsize=(9, 4))
plt.barh(inq_sorted['inquiry_type'], inq_sorted['count'], color='#9b59b6', height=0.55)
plt.title('Customer Inquiries by Type (Horizontal Bar)')
plt.xlabel('Inquiry Count')
plt.ylabel('Inquiry Type')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.show()

print("정렬된 문의 건수:")
print(inq_sorted)
````

**결과 해석 / 근거 / 새로 알게 된 점:**

1. `ascending=True`로 정렬하여 `plt.barh`를 실행한 결과, 맨 위에 가장 건수가 많은 **`상품 배송 일정 및 지연 문의(45건)`**가, 맨 아래에 **`회원 등급 및 포인트 적립 문의(20건)`**가 배치되었습니다.
2. 문의 유형과 같이 긴 텍스트 문장은 수직 막대에서는 라벨이 겹치거나 잘리지만, 수평 막대에서는 왼쪽 Y축에 좌측 정렬되어 아무리 긴 한글 문장도 깨끗하게 읽을 수 있습니다.

## 문제 7. `sns.countplot`으로 사전 집계 없이 빈도 비교하기 [선택 도전]

`online` 데이터에 대해 `sns.countplot(data=online, x='region', hue='payment_method')`를 실행하여 지역별·결제수단별 주문 건수를 시각화하세요.

**점검 기준:** 별도의 `groupby` 없이도 `South` 지역에서 `Card`가 4건, `Wallet`이 1건인 것 등 지역 내 결제수단별 건수가 자동 집계되어 표시됩니다.

````python
# 문제 7 풀이
plt.figure(figsize=(8, 4))
sns.countplot(
    data=online,
    x='region',
    hue='payment_method',
    palette='Set2'
)
plt.title('Online Order Counts by Region and Payment Method')
plt.xlabel('Region')
plt.ylabel('Order Count')
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.show()

print("South 지역 결제수단 분포:")
print(online[online['region'] == 'South']['payment_method'].value_counts())
````

## 문제 9. 상품 카테고리별 온라인·오프라인 평균 매출 그룹 막대그래프 [선택 도전]

`online` 복사본에 `channel='Online'`, `offline` 복사본에 `channel='Offline'` 열을 추가해 `pd.concat`으로 합친 뒤, X축 `product_category`, Y축 `revenue`, `hue='channel'`, `errorbar=None`으로 그룹 막대그래프를 그리세요.

**점검 기준:** 카테고리별 평균 매출은 `Tech`(Online `410.0` vs Offline `306.67`), `Book`(Online `133.33` vs Offline `170.0`), `Food`(Online `87.5` vs Offline `120.0`)로 `Tech`는 온라인이, `Book`·`Food`는 오프라인 평균 단가가 더 높습니다.

````python
# 문제 9 풀이
online_ch = online.copy()
online_ch['channel'] = 'Online'

offline_ch = offline.copy()
offline_ch['channel'] = 'Offline'

all_channel = pd.concat([online_ch, offline_ch], ignore_index=True)

plt.figure(figsize=(8, 4))
sns.barplot(
    data=all_channel,
    x='product_category',
    y='revenue',
    hue='channel',
    errorbar=None,
    palette='Set1'
)
plt.title('Average Revenue by Category: Online vs Offline')
plt.xlabel('Product Category')
plt.ylabel('Average Revenue')
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.show()

avg_comp = all_channel.groupby(['product_category', 'channel'])['revenue'].mean().unstack()
print("카테고리별 채널 평균 매출:")
display(avg_comp)
````

**결과 해석 / 근거 / 새로 알게 된 점:**

1. **카테고리별 채널 단가 비교:**
   - `Tech`: Online 평균 **410.00원** vs Offline 평균 **306.67원** (온라인이 약 103원 더 높음)
   - `Book`: Online 평균 **133.33원** vs Offline 평균 **170.00원** (오프라인이 약 37원 더 높음)
   - `Food`: Online 평균 **87.50원** vs Offline 평균 **120.00원** (오프라인이 약 32원 더 높음)
2. 고가 전자기기는 온라인 전문몰에서 고급 사양을 구매하는 경향이 있는 반면, 도서나 식품은 오프라인 매장 방문 시 묶음 구매 등으로 건당 결제액이 더 높게 나타나는 채널별 소비 특성을 확인할 수 있습니다.


## 문제 10. `sns.barplot`의 에러바(Error Bar) 유무 비교와 해석 [선택 도전]

`1행 2열` 서브플롯에 `online`의 `product_category`별 `revenue` 평균 막대를 왼쪽은 기본 설정(`errorbar=('ci', 95)`), 오른쪽은 `errorbar=None`으로 그리고 에러바가 의미하는 바를 적으세요.

**점검 기준:** 왼쪽 차트의 검은색 세로선은 95% 신뢰구간을 뜻하며, 표본 수가 적거나 값 편차가 큰 범주일수록 에러바가 길게 나타나 평균 추정의 불확실성이 큼을 알려줍니다.

**용어:** 신뢰구간은 같은 표집·계산 절차를 반복할 때 모평균을 포함하는 비율을 95%로 맞춘 구간입니다. 이 실습의 작은 가상 표본으로 실제 고객 전체의 평균을 입증하지 않습니다. 재현하려면 barplot에 seed=42를 지정합니다.

````python
# 문제 10 풀이
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

# 왼쪽: 에러바 있음 (95% 신뢰구간)
sns.barplot(data=online, x='product_category', y='revenue', errorbar=('ci', 95), ax=axes[0], palette='Blues_r')
axes[0].set_title('With Error Bar (95% CI)')
axes[0].set_xlabel('Product Category')
axes[0].set_ylabel('Average Revenue')
axes[0].grid(axis='y', linestyle='--', alpha=0.4)

# 오른쪽: 에러바 없음
sns.barplot(data=online, x='product_category', y='revenue', errorbar=None, ax=axes[1], palette='Blues_r')
axes[1].set_title('Without Error Bar')
axes[1].set_xlabel('Product Category')
axes[1].set_ylabel('Average Revenue')
axes[1].grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
````

**결과 해석 / 근거 / 새로 알게 된 점:**

### 에러바(Error Bar)의 통계적 의미와 필요성
1. **신뢰구간(Confidence Interval) 표시:** 왼쪽 차트 막대 상단의 검은색 수직선은 모평균이 존재할 95% 신뢰구간을 나타냅니다.
2. **해석상 위험 방지:**
   - 에러바가 없는 오른쪽 차트만 보면 `Tech` 카테고리의 평균이 410원으로 단정적으로 인식됩니다.
   - 그러나 에러바가 있는 왼쪽 차트를 보면 편차가 크고 표본 수가 5건으로 작아 신뢰구간 폭이 매우 넓음을 알 수 있습니다.
   - 즉, 에러바는 표본 평균 추정의 **통계적 불확실성(Uncertainty)**을 정직하게 전달하여 과도한 확증 편향을 막아주는 핵심 안전장치입니다.


## 문제 11. 피벗 테이블과 누적 막대그래프(`stacked=True`) [선택 도전]

문제 9에서 합친 전체 주문 데이터(`all_channel`)를 `index='product_category'`, `columns='channel'`, `values='revenue'`, `aggfunc='sum'`으로 피벗한 뒤 `.plot(kind='bar', stacked=True)`로 누적 막대그래프를 그리세요.

**점검 기준:** 각 카테고리 막대의 전체 높이는 총매출(`Tech 2970`, `Food 710`, `Book 740`)을 나타내고, 막대 내부 색상 분할로 `Online`과 `Offline`의 매출 기여 비중을 동시에 보여줍니다.

````python
# 문제 11 풀이
pivot_channel = all_channel.pivot_table(
    index='product_category',
    columns='channel',
    values='revenue',
    aggfunc='sum'
)

ax = pivot_channel.plot(kind='bar', stacked=True, figsize=(7, 4), color=['#e74c3c', '#3498db'], edgecolor='white')
plt.title('Category Revenue by Channel (Stacked Bar)')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.legend(title='Channel')
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.show()

print("피벗 테이블 (매출 합계):")
display(pivot_channel)
print("카테고리별 총매출:", pivot_channel.sum(axis=1).to_dict())
````

**결과 해석 / 근거 / 새로 알게 된 점:**

1. **누적 막대 차트 분석:**
   - 막대 전체 높이(총매출): `Tech` **2,970원**(Online 2,050 + Offline 920), `Book` **740원**(Online 400 + Offline 340), `Food` **710원**(Online 350 + Offline 360).
2. 카테고리별 전체 매출 규모(1위 Tech의 압도적 비중)와 채널별 기여도(Tech는 온라인 기여도가 지배적, Food와 Book은 온·오프라인 균등 기여)를 단 하나의 차트에서 입체적으로 읽어낼 수 있습니다.


## 문제 12. 결제수단별 매출 비중 파이 차트(`plt.pie`) [선택 도전]

`online`에서 `payment_method`별 `revenue` 합계를 구한 뒤, `plt.pie(..., labels=..., autopct='%.1f%%', explode=[0.05, 0, 0])`로 원형 차트를 그리고 각 결제수단의 점유율(%)을 확인하세요.

**점검 기준:** 온라인 총매출 `2800` 중 `Card`가 `2070(73.9%)`, `Wallet`이 `570(20.4%)`, `COD`가 `160(5.7%)`을 차지합니다.

````python
# 문제 12 풀이
pm_rev = online.groupby('payment_method')['revenue'].sum().loc[['Card', 'Wallet', 'COD']]

plt.figure(figsize=(6, 6))
plt.pie(
    pm_rev,
    labels=pm_rev.index,
    autopct='%.1f%%',
    explode=[0.05, 0, 0],
    colors=['#3498db', '#2ecc71', '#e74c3c'],
    startangle=140,
    wedgeprops=dict(edgecolor='white', linewidth=1.5)
)
plt.title('Online Revenue Share by Payment Method')
plt.show()

total_online_rev = pm_rev.sum()
for pm, val in pm_rev.items():
    print(f"{pm:7s}: {val:,}원 ({val/total_online_rev*100:.1f}%)")
print(f"전체 합계: {total_online_rev:,}원")
````

**결과 해석 / 근거 / 새로 알게 된 점:**

1. **온라인 결제수단 점유율 검증:**
   - 전체 온라인 매출: **2,800원**
   - `Card`: **2,070원 (73.9%)**
   - `Wallet`: **570원 (20.4%)**
   - `COD`: **160원 (5.7%)**
2. 1위 결제수단인 `Card`가 전체 매출의 3/4가량을 독점하고 있으며, `explode=[0.05, 0, 0]`을 통해 가장 중요한 카드 부채꼴을 돌출시켜 보고서 전달력을 높였습니다.
