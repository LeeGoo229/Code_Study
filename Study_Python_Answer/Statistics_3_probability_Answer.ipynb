## 문제 1. 표본공간·사건·조건부확률

공정한 주사위에서 A=짝수, B=4이상일 때 P(A), P(B), P(A∩B), P(A|B)를 구하고 독립인지 판단하세요.

**점검 기준:** 요구한 계산값·그래프와 사용한 가정, 단위를 함께 기록하세요.

```python
# 문제 1 풀이
s=set(range(1,7));a={2,4,6};b={4,5,6}
pa,pb,pab=len(a)/6,len(b)/6,len(a&b)/6
print(pa,pb,pab,pab/pb,'독립',np.isclose(pab,pa*pb))
```

**결과 해석 / 근거 / 새로 알게 된 점:**

각각 1/2,1/2,1/3,2/3입니다. 1/3≠1/4이므로 독립이 아닙니다.

## 문제 2. 이산균등분포와 기대값

값 [0,1,2]의 확률이 [.2,.5,.3]인 분포의 기대값과 분산을 직접 계산하세요.

**점검 기준:** 요구한 계산값·그래프와 사용한 가정, 단위를 함께 기록하세요.

```python
# 문제 2 풀이
v=np.array([0,1,2]);p=np.array([.2,.5,.3]);mu=sum(v*p);var=sum((v-mu)**2*p)
print(mu,var);assert np.allclose([mu,var],[1.1,.49])
```

**결과 해석 / 근거 / 새로 알게 된 점:**

기대값 1.1, 분산 .49입니다. 여기서 확률 가중 분산은 표본분산이 아니므로 n−1로 나누지 않습니다.

## 문제 3. 이항분포

독립 고객 20명의 구매확률이 .1일 때 정확히 2명, 1명 이상 구매할 확률과 기대 구매자 수를 계산하세요.

**점검 기준:** 요구한 계산값·그래프와 사용한 가정, 단위를 함께 기록하세요.

```python
# 문제 3 풀이
print(stats.binom.pmf(2,20,.1),stats.binom.sf(0,20,.1),stats.binom.mean(20,.1))
manual=math.comb(20,2)*.1**2*.9**18
assert np.isclose(manual,stats.binom.pmf(2,20,.1))
```

**결과 해석 / 근거 / 새로 알게 된 점:**

정확히 2명 약 .2852, 1명 이상 약 .8784, 기대 인원 2명입니다. sf(0)는 P(X>0)로 누적확률의 보완입니다.

## 문제 4. 포아송분포

시간당 평균 4건인 문의가 포아송모형을 따른다면 30분간 0건, 3건 이상일 확률을 구하세요.

**점검 기준:** 요구한 계산값·그래프와 사용한 가정, 단위를 함께 기록하세요.

```python
# 문제 4 풀이
lam=4*.5
print(stats.poisson.pmf(0,lam),stats.poisson.sf(2,lam))
assert np.isclose(stats.poisson.pmf(0,lam),np.exp(-2))
```

**결과 해석 / 근거 / 새로 알게 된 점:**

30분의 λ는 2입니다. 확률은 약 .1353과 .3233입니다. 업무 시간별 발생률이 크게 다르면 일정률 가정을 다시 확인합니다.

## 문제 5. 연속분포와 구간 확률

배달시간이 정규모형 N(35,5²)일 때 30분 미만과 40분 이상 확률을 구하세요. 같은 평균이면 서비스 신뢰성이 같은지도 설명하세요.

**점검 기준:** 요구한 계산값·그래프와 사용한 가정, 단위를 함께 기록하세요.

```python
# 문제 5 풀이
print(stats.norm.cdf(30,35,5),stats.norm.sf(40,35,5))
assert np.isclose(stats.norm.cdf(30,35,5),stats.norm.sf(40,35,5))
```

**결과 해석 / 근거 / 새로 알게 된 점:**

각각 약 .1587입니다. 평균이 같아도 분산과 늦은 배달 꼬리 확률이 다르면 마감시간을 지킬 가능성도 달라집니다. 정규모형은 음수 시간도 허용하므로 실제 데이터 적합성을 확인해야 합니다.
