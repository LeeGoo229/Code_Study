# 1. 데이터 미리보기 및 구조 확인

```python
import pandas as pd

orders = pd.DataFrame({
    "order_id": [1001, 1002, 1003, 1004, 1005, 1006],
    "customer": ["민수", "지수", "철수", "영희", "준호", "수진"],
    "price": [15000, 32000, 8000, 45000, 12000, 27000]
})

# 데이터 앞부분 확인
print(orders.head(3))
#    order_id customer  price

# 데이터 뒷부분 확인
print(orders.tail(2))
#    order_id customer  price

# 행과 열 개수 확인
print(orders.shape)

# 컬럼명 확인
print(orders.columns)

# 데이터 타입 확인
print(orders.dtypes)

# 컬럼명 변경
orders = orders.rename(columns={"price": "order_price"})
```

| 메서드 | 설명 |
| --- | --- |
| `df.shape` | (행 수, 열 수) 튜플 반환 |
| `df.dtypes` | 각 컬럼의 데이터 타입 확인 |
| `df.info()` | 열별 비결측값 수·자료형·메모리 정보. 결측 수는 전체 행 수에서 비결측 수를 빼서 계산 |
| `df.columns` | 컬럼명 확인 |
| `df.rename()` | 컬럼명 변경 |

### 구조를 읽는 순서

`head()`와 `tail()`의 기본값은 5행입니다. `shape`, `columns`, `dtypes`는 속성이므로 괄호를 붙이지 않습니다. `info()`는 정보를 출력하고 `None`을 반환하므로 `df.info()`로 호출합니다. 큰 데이터에서는 `df.info(show_counts=True)`로 비결측값 수를 명시할 수 있습니다.

`rename()`은 기본적으로 변경된 DataFrame을 반환하므로 결과를 변수에 다시 저장합니다. 한 열을 `df["price"]`로 선택하면 Series, `df[["price"]]`로 선택하면 DataFrame입니다.

# 2. 데이터 타입 확인과 변환

```python
import pandas as pd

products = pd.DataFrame({
    "product": ["키보드", "마우스", "모니터"],                   # str
    "price": ["45000", "25000", "300000"],                      # str
    "sale_date": ["2026-08-01", "2026-08-03", "2026-08-05"]     # str
})
print(products.dtypes)

# 문자열을 분석에 맞는 자료형으로 변환
products["price"] = products["price"].astype(int)               
products["sale_date"] = pd.to_datetime(products["sale_date"])   

print(products.dtypes)

# price 평균 값 출력
print(products["price"].mean())
```

| 타입 | 설명 |
| --- | --- |
| int64 | 정수형 |
| float64 | 실수형 |
| object | 문자열 등을 담을 수 있는 일반 객체형. 문자열 전용 string/str 자료형도 사용됨 |
| bool | True/False |
| datetime64 | 날짜/시간 |
| category | 범주형 |

# 3. 데이터 분포 파악

```python
import pandas as pd

orders = pd.DataFrame({
    "user_id": [101, 102, 101, 103, 104, 102],
    "category": ["식품", "의류", "식품", "전자기기", "의류", "식품"],
    "price": [12000, 45000, 18000, 300000, 55000, 15000]
})

# 수치형 데이터의 통계 요약
print(orders["price"].describe())

# 카테고리별 주문 횟수
print(orders["category"].value_counts())

# 실제 주문 고객 수
print(orders["user_id"].nunique())
```

**describe(): 수치형 열 통계 요약**

`describe()`는 수치형 컬럼에 대한 기본 통계량을 종합하여 보여줍니다.

| 통계량 | 설명 | 눈여겨봐야 할 포인트 |
| --- | --- | --- |
| count | 누락값 제외한 데이터 수 | 전체 행 수와 비교해서 count가 적다면 결측치가 있다는 신호 |
| mean | 평균값 | 평균과 중앙값 차이는 치우침을 의심할 단서이며 분포를 추가 확인  |
| std | 표준편차 | 변동성, 표준편차가 평균보다 크면 이상치나 이질적인 그룹이 섞여 있을 가능성이 있음  |
| min / max | 최소/최댓값 | 물리적으로 불가능한 값이 있는지  |
| 25%, 50%, 75% | 사분위수 | 분포의 위치·범위 파악. 이 값만으로 이상치를 확정하지 않음 |


**value_counts(), nunique(), corr()**

범주형과 수치형 변수의 분포를 파악하는 메서드들입니다.

| 메서드 | 설명 |
| --- | --- |
| `value_counts()` | 범주형 컬럼의 개별 값 빈도 출력 |
| `nunique()` | 유일값 개수 (카디널리티 확인) |
| `corr()` | 수치형 변수 간 상관계수 행렬 (시각화 챕터 예고) |


### 통계를 해석할 때

`describe()`의 `count`는 비결측값 수이고, `std`는 기본적으로 표본 표준편차(ddof=1)입니다. `50%`는 중앙값입니다. 평균이 중앙값보다 크면 큰 값들의 영향을 의심할 수 있지만, 이것만으로 오류나 이상치를 확정하지 않습니다.

`count()`는 비결측값 개수, `nunique()`는 서로 다른 비결측값 개수입니다. `value_counts()`와 `nunique()`는 기본적으로 결측값을 제외하며 `dropna=False`로 포함할 수 있습니다. 각 행이 주문 한 건이라는 전제에서만 범주별 행 빈도를 주문 건수로 해석합니다.

`corr()`는 숫자 열을 명시적으로 선택해 사용합니다. 상관관계만으로 인과관계를 판단하지 않습니다.

### 자료형 변환 주의점

`astype(int)`는 변환 가능한 값에 사용합니다. 문자·소수·누락값이 섞여 있으면 먼저 확인해야 합니다. 변환할 수 없는 값을 누락값으로 표시하려면 `pd.to_numeric(..., errors="coerce")`를 사용할 수 있으나, 변환 전후 결측 수를 비교해야 합니다. 누락값이 있는 정수 열은 pandas의 `Int64` 자료형을 고려합니다. 날짜는 `pd.to_datetime()`으로 변환하며 변환 실패를 `NaT`로 처리할 경우 그 건수를 확인합니다. 날짜 시간 단위와 문자열 자료형 표시는 pandas 버전에 따라 달라질 수 있습니다.

# 4. 데이터 선택과 필터링

| 방식 | 기준 | 범위 끝 처리 |
| --- | --- | --- |
| `iloc` | 0부터 시작하는 정수 위치 | 끝 위치 제외 |
| `loc` | 행 인덱스 레이블·열 이름 | 레이블 슬라이스 끝 포함 |
| Boolean Indexing | 조건식의 True/False | True인 행 선택 |

```python
customers = pd.DataFrame({"name": ["민수", "지수", "철수", "영희", "준호"],
                          "region": ["서울", "부산", "서울", "서울", "대전"],
                          "point": [1500, 4200, 3500, 5000, 2800]})
customers.iloc[:3, [0, 2]]              # 앞 3행, 위치 0·2의 열
customers.loc[:, ["name", "point"]]     # 모든 행에서 두 열
customers.loc[(customers["region"] == "서울") & (customers["point"] >= 3000),
              ["name", "point"]]
```

`iloc[5:10]`은 위치 5~9, 즉 여섯 번째부터 열 번째 행입니다. `loc[0:4]`는 인덱스 레이블이 0~4일 때 해당 레이블 범위를 선택하므로, 인덱스가 바뀌면 처음 5행과 다를 수 있습니다.

여러 조건에는 `&`(그리고), `|`(또는)를 쓰고 각각 괄호로 감쌉니다. pandas Series 조건에 Python의 `and`, `or`를 쓰면 오류가 발생합니다.
